**TASK 2 :**
2) Add New Job Postings (interactive)
Ask the user how many job postings they want to add. For each posting, collect the role,
company, and skills (skills provided as a single string, space-separated or comma-separated).
Requirements:
● Job_ID should auto-increment from the maximum existing ID.
● Convert numeric inputs (if any) to integers where applicable.
● Append new values to the lists in job_data.

Example interaction
How many job postings do you want to add? 2
Enter role: Junior Data Analyst
Enter company: BrightData
Enter skills (space-separated or comma-separated): Excel, SQL, Python

```python

# Ask how many job postings the user wants to add
n = int(input("How many job postings do you want to add? "))

# Find the maximum existing Job_ID
max_id = max(job_data["Job_ID"])

# Add new job postings
for i in range(n):
    role = input("Enter role: ")
    company = input("Enter company: ")
    skills = input("Enter skills (space-separated or comma-separated): ")

    # Generate automatic Job_ID
    new_id = max_id + 1

    # Append new details to dictionary
    job_data["Job_ID"].append(new_id)
    job_data["Role"].append(role)
    job_data["Company"].append(company)
    job_data["Skills"].append(skills)

    max_id = new_id

# Display updated job data
print("\nUpdated Job Data:")
print(job_data)
```

**TASK:3**
3) Clean & Normalize Skills (text processing)
Make a function called clean_skills() that accepts skills in string format.
● Converts text to lowercase
● Replaces commas with spaces
● Removes extra spaces (normalize whitespace)
● Optionally removes punctuation (you may use basic str.replace() or re module)
Apply clean_skills() to every entry in job_data['Skills'] and overwrite the list with cleaned
versions.

```python
# Function to clean skills
import re

def clean_skills(skills):
    skills = skills.lower()               # Convert to lowercase
    skills = skills.replace(",", " ")     # Replace commas with spaces
    skills = re.sub(r'[^\w\s]', '', skills)  # Remove punctuation
    skills = " ".join(skills.split())     # Remove extra spaces
    return skills

# Apply function to every skill in the list
for i in range(len(job_data["Skills"])):
    job_data["Skills"][i] = clean_skills(job_data["Skills"][i])

# Display cleaned skills
print("Cleaned Skills:")
for i in range(len(job_data["Skills"])):
    print(job_data["Skills"][i])
```

**TASK 4 :**
4) Skill Extraction & Frequency
● From the cleaned skills list, split each skill string into individual skill tokens.
● Build a consolidated list of all skills that contains every skill token across all postings.
● Use a dictionary or other structure to compute skill_count = {skill: frequency}.
● Print the top 5 most frequent skills (or all if < 5).
Expected print example:
Top skills by frequency:
Python: 5
sql: 4
Excel: 2
ML: 2
Power BI: 1

```python
# Create an empty list to store all skills
all_skills = []

# Extract individual skills from each job posting
for i in range(len(job_data["Skills"])):
    skills = job_data["Skills"][i].split()

    for skill in skills:
        all_skills.append(skill)

# Calculate skill frequency
skill_count = {}

for skill in all_skills:
    if skill in skill_count:
        skill_count[skill] += 1
    else:
        skill_count[skill] = 1

# Sort skills based on frequency (highest first)
sorted_skills = sorted(
    skill_count.items(),
    key=lambda x: x[1],
    reverse=True
)

# Print top 5 most frequent skills
print("Top skills by frequency:")

for skill, count in sorted_skills[:5]:
    print(skill + ":", count)
```

**TASK:5**
5) Basic Statistics & Insights
Calculate and print:
● Total number of job postings
● Number of unique skills
● Jobs that mention "python" (case-insensitive)
Sample Output
Total Jobs: 7
Unique skills: 9
Jobs mentioning Python: ['Data Analyst', 'Data Scientist', 'Data Engineer', ...]

```python

# 1. Calculate total number of job postings
total_jobs = len(job_data["Job_ID"])

print("Total Jobs:", total_jobs)


# 2. Calculate number of unique skills
unique_skills = set()

for i in range(len(job_data["Skills"])):
    skills = job_data["Skills"][i].split()

    for skill in skills:
        unique_skills.add(skill.lower())

print("Unique skills:", len(unique_skills))


# 3. Find jobs mentioning Python
python_jobs = []

for i in range(len(job_data["Skills"])):
    if "python" in job_data["Skills"][i].lower().split():
        python_jobs.append(job_data["Role"][i])

print("Jobs mentioning Python:", python_jobs)
```

**TASK:6**
6) Skill Search Function (functions & conditional logic)
Create a function, find_jobs_by_skill(skill_query), that returns a list of (Job_ID, Role,
Company) for postings containing the skill_query token.
● The function must normalize the skill_query (lowercase, strip spaces)
● Use if/elif/else where appropriate
Example: find_jobs_by_skill('sql') → returns matching jobs.

```python

# Function to search jobs by skill
def find_jobs_by_skill(skill_query):

    # Normalize the search skill
    skill_query = skill_query.lower().strip()

    matching_jobs = []

    for i in range(len(job_data["Skills"])):

        skills = job_data["Skills"][i].split()

        if skill_query in skills:
            matching_jobs.append((
                job_data["Job_ID"][i],
                job_data["Role"][i],
                job_data["Company"][i]
            ))

        elif skill_query == "":
            print("Please enter a valid skill.")
            return []

        else:
            continue

    return matching_jobs


# Ask user for a skill to search
search_skill = input("Enter a skill to search: ")

# Call the function
result = find_jobs_by_skill(search_skill)

# Display matching jobs
if result:
    print("\nJobs matching", search_skill, ":")

    for job in result:
        print("Job ID:", job[0],
              "| Role:", job[1],
              "| Company:", job[2])

else:
    print("No jobs found for this skill.")
```

**TASK:7**
7) Lambda Filtering & List Comprehension
Use a lambda function and filter() (or a list comprehension with a lambda-like expression) to
extract job roles where the skill count (from skill_count) for any skill >= 2 — that is, roles that
include at least one relatively common skill.
Example Output
Roles with at least one common skill: ['Data Analyst', 'Data Scientist', 'ML Engineer']

```python

# Use lambda function and filter() to find roles
common_skill_roles = list(filter(
    lambda i: any(
        skill_count[skill] >= 2
        for skill in job_data["Skills"][i].split()
    ),
    range(len(job_data["Role"]))
))

# Extract matching job roles
common_roles = [
    job_data["Role"][i] for i in common_skill_roles
]

# Display the result
print("Roles with at least one common skill:", common_roles)
```
